# P9 Optical Anomaly Patterns Analysis
## Lakelink Fiber — Provable Resilience Intelligence

**Task:** P9-OPTICAL-ANOMALY-PATTERNS — Analyze optical degradation patterns in 5M telemetry rows.

### Key Findings

**1. OSNR-BER Threshold:** Clear nonlinear relationship. BER breach rate jumps from 0.3% at OSNR 21 dB to 8% at OSNR 19 dB. Critical threshold: **20 dB**. Below 17 dB: 15-16% breach rate.

**2. BER Breach Precursors:** No significant OSNR drop in 6h before BER breach (avg OSNR ~24.6 across all pre-breach windows). BER breaches are SUDDEN events, not preceded by gradual OSNR degradation.

**3. Multi-Span Correlated Degradation:** 2,160 events where ≥5 spans degrade in the same hour. Average 106 spans per event (max 141). Mean OSNR during: 17.28 dB. Confirms shared amplifier/DWDM chain failure mode.

**4. Seasonal Patterns:** No meaningful temporal patterns — breach rate uniform across hours (0.75-0.82%), days (0.78-0.81%), and weeks (OSNR stable at 24.59-24.60 dB). Synthetic data limitation.

**5. Span Trends:** 416 spans (4.1%) degrading, 399 (4.0%) improving, 9,185 (91.8%) stable.

**6. Feature Enhancement:** `osnr_weekly_trend` (importance 0.12) and `osnr_volatility` (0.10) are NEW features not in the current span_risk_classifier. Recommend adding as features for v4 model.
